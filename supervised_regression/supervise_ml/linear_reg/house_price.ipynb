```python
# df.isnull().sum()
# df.dtypes
# df.dropna()
df["SalePrice"] = df["SalePrice"].fillna(df["SalePrice"].mean())
df["MSZoning"] = df["MSZoning"].fillna(df["MSZoning"].mode()[0])
df["Exterior1st"] = df["Exterior1st"].fillna(df["Exterior1st"].mode()[0])
df[["BsmtFinSF2" , "TotalBsmtSF" ]]= df[["BsmtFinSF2","TotalBsmtSF"]].fillna(df [["BsmtFinSF2" ,"TotalBsmtSF"]].mean())
```

```python
df = pd.get_dummies(df , columns =["LotConfig" , "BldgType" ,"Exterior1st"] , dtype = "int")
```

```python
df.dtypes
```

```python
df = pd.get_dummies(df , columns = ["MSZoning"] , dtype = 'int')
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
x = df.drop(columns = ["SalePrice"])
y = df["SalePrice"]
X_train, X_test, y_train, y_test = train_test_split(
x, y, test_size=0.33, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
linear_pred = model.predict(X_test)
from sklearn.metrics import r2_score
r2_linear = r2_score(y_test ,linear_pred)
```
